**Tipos** de datos en Python **texto en negrita**

```python
# int
numero = 10

print(f"La variable numero tiene un {numero}  y es de tipo {type(numero).__name__}\n")

# str
nombre = 'Santiago'
print(f"La variable nombre tiene un {nombre}  y es de tipo {type(nombre).__name__}\n")

# bool
condicion = True
print(f"La variable condicion tiene un {condicion} y es de tipo {type(condicion).__name__}\n")

# float
numero_decimal = 15.5
print(f"La variable numero_con_coma tiene un {numero_decimal} y es de tipo {type(numero_decimal).__name__}\n")

# complex
complejo = 1 + 2j
print(f"La variable complejo tiene un {complejo} y es de tipo {type(complejo).__name__}\n")
```

**Tipos de datos coleccionesd**

```python
# Listas
lista = [1,2,3,4,5,6]
print(f"La variable lista tiene un {lista} y es de tipo {type(lista).__name__}\n")

# Tuplas, Son como las listas pero no se pueden modificar
tupla = (1,2,3,4,5)
print(f"La variable tupla tiene un {tupla} y es de tipo {type(tupla).__name__}\n")
tupla_un_elemento = (6,)
print(f"La variable tupla_un_elemento tiene un {tupla_un_elemento} y es de tipo {type(tupla_un_elemento).__name__}\n")


# Diccionarios (Como un jsson)
persona = {
    "nombre" : "Juan",
    "apellido" : "Perez",
    "edad" : 25
}
print(f"La variable persona tiene un {persona} y es de tipo {type(persona).__name__}\n")


# set
pares_del_1_al_10 = {2,4,6,8,10}
print(f"La variable pares_del_1_al_10 tiene un {pares_del_1_al_10} y es de tipo {type(pares_del_1_al_10).__name__}\n");
if 2 in pares_del_1_al_10:
    print("El numero 2 esta en el set\n")

# range
# Defino un rango de 1 a 10

rango = range(1,11)
print(f"La variable rango tiene un {rango} y es de tipo {type(rango).__name__}")
for numero in rango:
    print(numero)

# Ejemplo si quiero iterar los numeros de 2 en 2

print("------------------------------------")
for numero in range(1,11,2):
    print(numero)

# Recorrer strings

print("------------------------------------")
cadena = "Hola"
for letra in cadena:
    print(letra)

print("------------------------------------")
lista = [1,2,3,4,5,6]
for numero in lista:
    print(numero)
```

```python
import pandas as pd  # requires: pip install 'pandas[pyarrow]'
from chronos import Chronos2Pipeline
pipeline = Chronos2Pipeline.from_pretrained("amazon/chronos-2")
```

```python
import numpy as np
datos = np.array([1,2,3,4,5,6,7], dtype=np.float32)
prediccion  = pipeline.predict([datos], prediction_length=1)
print(prediccion[0].mean())
```
